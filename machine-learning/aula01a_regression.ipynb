````python
# 1. Extrai todas as linhas (:) da coluna de índice 2, mas usando 2:3 garantimos que continue sendo uma matriz 2D
# Guardamos esse resultado na nova variável 'X2'
X2 = X[:,2:3]

# 2. Imprimimos o formato de X2. O resultado deve ser (442, 1), que é uma matriz de 1 coluna
print(X2.shape)
````

#### 🔍 O que este bloco faz?
Ele isola a coluna do IMC e a transforma em uma **nova matriz bidimensional** chamada `X2`, verificando em seguida se ela manteve o formato adequado `(442, 1)`.

#### 💡 Dica Técnica:
Por que usar `X[:, 2:3]` em vez de apenas `X[:, 2]`?
* `X[:, 2]` retorna um vetor de uma dimensão: formato `(442,)`.
* `X[:, 2:3]` retorna uma matriz coluna bidimensional: formato `(442, 1)`.
* A maioria dos modelos de *Scikit-Learn* exige que as entradas (`X`) sejam estritamente bidimensionais, mesmo que você use apenas uma característica para treinar o modelo.


````python
# 1. Define uma nova função chamada 'modelo_linear' que recebe as entradas X, uma inclinação 'a' e uma altura 'b'
def modelo_linear(X, a, b):

  # 2. Aplica a fórmula matemática da equação da reta: y = a * X + b
  # Isso calcula a nossa 'previsão' para cada valor de X
  y = a*X + b

  # 3. Devolve (retorna) os valores previstos
  return y
````

````python
# 1. Define uma função chamada 'plot_results' para facilitar o desenho dos gráficos
# Ela recebe os dados de entrada (X), os valores reais (y) e as nossas previsões (ypred)
def plot_results(X, y, ypred):

  # 2. Plota os pontos reais (X vs y) como bolinhas azuis
  plt.scatter(X, y)

  # 3. Plota a linha das nossas previsões (X vs ypred). O 'r' significa 'red' (vermelho)
  plt.plot(X, ypred, 'r')

  # 4. Exibe o gráfico na tela
  plt.show()
````

#### 🔍 O que estes blocos fazem? (modelo_linear e plot_results)
O professor cria duas funções personalizadas no Python:
1. **`modelo_linear(X, a, b)`:** É a representação pura da **Equação da Reta** da matemática do ensino médio ($y = aX + b$). Onde `a` é a inclinação da reta (peso matemático) e `b` é onde ela cruza o eixo Y (viés ou *bias*).
2. **`plot_results(X, y, ypred)`:** Uma função para facilitar o desenho gráfico. Ela desenha as bolinhas reais (`y`) em azul e a linha reta das nossas previsões (`ypred`) em vermelho ('r').

#### 🎯 Qual a intenção pedagógica?
Você está construindo o modelo de Regressão Linear mais primitivo possível: **na mão!** Antes de usar inteligências artificiais prontas, você precisa entender que um modelo linear é, no fundo, apenas encontrar os melhores valores para `a` e `b` que façam a reta vermelha passar exatamente pelo meio das bolinhas azuis.

````python
# 1. Usamos a nossa função da equação da reta chutando os valores a=1000 e b=100.
# Guardamos o resultado dessas previsões na variável 'ypred'
ypred = modelo_linear(X2, 1000, 100)

# 2. Desenhamos o gráfico comparando a realidade com as nossas previsões
plot_results(X2, y, ypred)
````

````python
# 1. Tentamos um novo chute, agora com a=1500 e b=150, tentando ajustar melhor a reta.
# Guardamos as novas previsões na variável 'ypred'
ypred = modelo_linear(X2, 1500, 150)

# 2. Desenhamos o gráfico novamente para ver se a reta vermelha melhorou
plot_results(X2, y, ypred)
````

#### 🔍 O que estes blocos fazem? (Chute inicial)
Nestas células, o professor está definindo manualmente os valores de `a` e `b` e plotando os resultados.

#### 🎯 Qual a intenção pedagógica?
Mostrar o processo de **tentativa e erro**.
* No primeiro teste (`a=1000`, `b=100`), a reta vermelha provavelmente não cruzou a nuvem de pontos da melhor forma.
* No segundo teste (`a=1500`, `b=150`), ele ajusta os parâmetros da reta (alterando a inclinação e altura) para tentar encaixá-la melhor nos dados reais.
* É exatamente isso que os algoritmos de Machine Learning fazem: eles testam sistematicamente e de forma inteligente milhares de valores para `a` e `b` até acharem a reta perfeita (o menor erro).


````python
# Importa a biblioteca numpy (apelidada de 'np'), que é uma ferramenta super poderosa para cálculos matemáticos
import numpy as np
````

#### 🔍 O que os próximos blocos fazem? (Métricas de Erro)
O professor cria manualmente três funções para calcular o quão "ruim" foi o nosso chute (o quão distante a reta vermelha passou dos pontos azuis reais).

#### 🎯 Qual a intenção pedagógica?
Apresentar as três métricas de erro mais utilizadas no mundo para problemas de **Regressão**:
1. **MAE (Erro Médio Absoluto):** Pega a distância de cada ponto para a reta, transforma os valores negativos em positivos (módulo absoluto) e tira a média. É muito fácil de interpretar (ex: "erramos o peso em média por 5kg").
2. **MSE (Erro Quadrático Médio):** Em vez de pegar o valor absoluto, eleva a diferença ao quadrado. Isso faz com que erros muito grandes sejam "punidos" severamente pelo algoritmo. É a métrica mais usada pelos algoritmos para treinar o modelo.
3. **RMSE (Raiz do Erro Quadrático Médio):** Como o MSE eleva tudo ao quadrado, o valor final perde a unidade de medida original (ficaria algo como $kg^2$). Tirando a raiz quadrada (RMSE), o valor volta para a escala original dos dados, mas mantendo a punição para erros grandes.

> [!TIP]
> O objetivo de todo algoritmo de Machine Learning de Regressão é encontrar matematicamente a reta que resulte no **menor valor possível** para essas métricas de erro.

````python
# Métrica 1: MAE (Erro Médio Absoluto)

# Define a função mae que recebe os valores reais (y) e os valores previstos (ypred)
def mae(y, ypred):
  # 1. Subtrai o valor real do previsto (calcula a distância)
  diff = y - ypred.reshape(len(y), )
  # 2. Pega o valor absoluto (tira o sinal negativo de quem ficou abaixo da reta)
  abs_diff = abs(diff)
  # 3. Soma todos os erros
  sum_error = np.sum(abs_diff)
  # 4. Divide pela quantidade total de pontos para ter a média
  mae = sum_error / len(y)
  # 5. Retorna o valor do Erro Médio Absoluto
  return mae

# Testamos e comparamos o erro do nosso primeiro chute com o do nosso segundo chute
mae(y, modelo_linear(X2, 1000, 100)), mae(y, modelo_linear(X2, 1500, 150))
````

````python
# Métrica 2: MSE (Erro Quadrático Médio)

# Define a função mse que recebe os valores reais (y) e os valores previstos (ypred)
def mse(y, ypred):
  # 1. Subtrai o valor real do previsto (calcula a distância)
  diff = y - ypred.reshape(len(y), )
  # 2. Eleva essa diferença ao quadrado (para tirar os negativos e punir erros grandes)
  squared = diff**2
  # 3. Calcula a média de todos esses quadrados
  mean = sum(squared)/len(squared)
  # 4. Retorna o valor do Erro Quadrático Médio
  return mean

# Testamos e comparamos o erro do nosso primeiro chute com o do nosso segundo chute
mse(y, modelo_linear(X2, 1000, 100)), mse(y, modelo_linear(X2, 1500, 150))
````

````python
# Métrica 3: RMSE (Raiz do Erro Quadrático Médio)

# Define a função rmse que recebe os valores reais (y) e os valores previstos (ypred)
def rmse(y, ypred):
  # Simplesmente calcula a raiz quadrada (np.sqrt) do resultado que a nossa função MSE (acima) deu
  return np.sqrt(mse(y, ypred))

# Testamos e comparamos o erro do nosso primeiro chute com o do nosso segundo chute
rmse(y, modelo_linear(X2, 1000, 100)), rmse(y, modelo_linear(X2, 1500, 150))
````
